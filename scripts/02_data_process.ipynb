## LUAD DGE Data Process with Raw Counts

```python
## import the packages

import pandas as pd 
import numpy as np
import os
```

```python
DGE_PATH = "../data/LUAD_MERGED_ALL_DEG_RESULT.csv"
df = pd.read_csv(DGE_PATH)
```

```python
PVAL_COL = [col for col in df.columns if col.endswith("_P_VALUE")]
select_col = ['ID'] + PVAL_COL
# filtered_df = df[select_col].loc[(df[PVAL_COL] <= 0.05).all(axis=1)]

df = df[select_col]

filtered_df = df[df[PVAL_COL].le(0.05).sum(axis=1) >= len(PVAL_COL)*0.9]
# filtered_df.to_csv("DATA/COMMON_SIGNIFICANT_GENES.csv", index = False)
filtered_df
```

```python

NORMAL_DIR = "../data/counts/NORMAL"
TUMOR_DIR = "../data/counts/TUMOR"

def MERGE_DFS(DIR_PATH):
    df_list = []
    for file in os.listdir(DIR_PATH):
        if file.endswith('.tsv'):
            df = pd.read_csv(os.path.join(DIR_PATH, file), sep= "\t")
            df_list.append(df)
    merged_df = pd.concat(df_list, axis=1)
    return merged_df

NORMAL_COUNT = MERGE_DFS(NORMAL_DIR)

TUMOR_COUNT = pd.read_csv(os.path.join(TUMOR_DIR, "Tumor_LUAD_Counts.tsv"), sep = "\t")

```

```python

NORMAL_FEA = NORMAL_COUNT.T
TUMOR_FEA = TUMOR_COUNT.T
NORMAL_FEA["Label"] = 0
TUMOR_FEA["Label"] = 1

DATA_MAT = pd.concat([TUMOR_FEA, NORMAL_FEA])

```

```python

SIG_GENES = list(filtered_df['ID'].values)

DATA_MAT_1 = DATA_MAT[SIG_GENES]
DATA_MAT_1["Label"] = DATA_MAT["Label"]

DATA_MAT_1.to_csv("../data/FEATURE_DATA_MATRIX.csv")
```
